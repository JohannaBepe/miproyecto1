```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('Titanic.csv')
df.head()


```

```python
df['TamañoFamilia'] = df['SibSp'] + df['Parch']+1
df.head()
```

```python
plt.figure (figsize=(10,6))
sns.countplot(x='Sobreviviente',hue='Sexo',data=df)
plt.title('Sobrevivientes por sexo')
plt.show()

plt.figure (figsize=(10,6))
sns.countplot(x='Sobreviviente',hue='Pclase',data=df)
plt.title('Sobrevivientes por clase')
plt.show()

plt.figure (figsize=(10,6))
sns.countplot(x='Sobreviviente',hue='Embarco',data=df)
plt.title('Sobrevivientes por puerta de embarque')
plt.show()

plt.figure (figsize=(10,6))
sns.countplot(x='Sobreviviente',hue='TamañoFamilia',data=df)
plt.title('Sobrevivientes por tamaño de la familia')
plt.show()
```

```python
Sobreviviente_by_Sexo = df.groupby('Sexo')['Sobreviviente'].sum()
plt.pie(Sobreviviente_by_Sexo,labels=Sobreviviente_by_Sexo.index,autopct='%1.1f%%')
plt.title('Propoción de sobrevivientes por sexo')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('Titanic.csv')
df.head()

Sobreviviente_by_Sexo = df.groupby('Sexo')['Sobreviviente'].sum()
plt.pie(Sobreviviente_by_Sexo,labels=Sobreviviente_by_Sexo.index,autopct='%1.1f%%')
plt.title('Proporción de sobrevivientes por sexo')
plt.show()

Sobreviviente_by_Clase = df.groupby('Pclase')['Sobreviviente'].sum()
plt.pie(Sobreviviente_by_Clase,labels=Sobreviviente_by_Clase.index,autopct='%1.1f%%')
plt.title('Proporción de sobrevivientes por clase')
plt.show()

```

```python
survived_by_sex = df.groupby('Sexo')['Sobreviviente'].sum()

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.pie(survived_by_sex, labels=survived_by_sex.index, autopct='%1.1f%%')
plt.title('Proporción de sobrevivientes por Sexo')

plt.subplot(1, 2, 2)
survived_by_sex.plot(kind='bar')
plt.title('Proporción de sobrevivientes por sexo')
plt.show()
```
